```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# =========================================================
# Assignment 1 & Mini Project 3: K-Means on Iris Dataset
# =========================================================
print("--- 1. K-MEANS CLUSTERING ON IRIS DATASET ---")

# Load Iris dataset
iris = load_iris()
X = iris.data
y_true = iris.target
feature_names = iris.feature_names

# Apply K-Means with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
y_pred = kmeans.fit_predict(X)

# =========================================================
# Assignment 2: Apply PCA to reduce dataset dimensions
# =========================================================
print("\n--- 2. DIMENSION REDUCTION USING PCA ---")
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print(f"Original shape: {X.shape}")
print(f"Reduced shape after PCA: {X_pca.shape}")
print(f"Explained variance ratio: {pca.explained_variance_ratio_}")

# =========================================================
# Mini Project 3: Visualizations & Cluster Comparison
# =========================================================
# Visualize PCA clusters vs True Labels side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Predicted Clusters Plot
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y_pred, cmap='viridis', edgecolor='k', s=50)
axes[0].set_title('Mini Project 3: K-Means Predicted Clusters (k=3)')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')
axes[0].grid(True, linestyle='--', alpha=0.6)

# True Labels Plot
axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=y_true, cmap='Set1', edgecolor='k', s=50)
axes[1].set_title('Mini Project 3: True Iris Flower Species Labels')
axes[1].set_xlabel('Principal Component 1')
axes[1].set_ylabel('Principal Component 2')
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.savefig('iris_clustering_pca.png', dpi=300, bbox_inches='tight')
plt.show()

# Compare predicted clusters vs true labels
print("\n--- 3. COMPARISON (Predicted Clusters vs True Labels) ---")
df_compare = pd.DataFrame({'True_Label': y_true, 'Cluster_Pred': y_pred})
print(pd.crosstab(df_compare['Cluster_Pred'], df_compare['True_Label']))
```
